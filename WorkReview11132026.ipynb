```python
import re

def get_color_mpl_new(color_in):
    trace_color = {}
    color_in = color_in.lower()

    valid_colors = COLORS
    pattern_list = [r"#[a-f0-9]{6}.[0-9]", r"[bgrcmykw].[0-9]", r"[bgrcmykw]", r"#[a-f0-9]{6}"]

    # first check if color is one of the matplotlib supported css colors
    if color_in in valid_colors:
        trace_color.update({'color':color_in, 'alpha':1})
        return trace_color

    valid = False
    for pattern in pattern_list:
        if re.fullmatch(pattern, color_in):
            valid = True

    if valid:
        color_split = color_in.split(".")
        color = color_split[0]
        # contains an alpha parameter
        if len(color_split) == 2:
            alpha = "0."+color_split[1]
            trace_color.update({'color':color, 'alpha':alpha})
        else:
            trace_color.update({'color':color, 'alpha':1})

        return trace_color

    else:
        trace_color.update({'color':'k', 'alpha':1})
        return trace_color

# example
color_in = "#3f0825.2"
get_color_mpl_new(color_in)
```
